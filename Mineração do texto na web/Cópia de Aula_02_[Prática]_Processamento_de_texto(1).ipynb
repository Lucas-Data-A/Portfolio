````python
import nltk
nltk.download('punkt_tab')
text = "This is a English sample of the lib NLTK. NLTK is a lib for NLP."
sentence_tokenizer = nltk.data.load('tokenizers/punkt/english.pickle')
result = sentence_tokenizer.tokenize(text)
print(len(result))

for n, sentence in enumerate(result):
  print(f"{n}: {sentence}")
````

````python
import nltk
nltk.download('mac_morpho')
sents = nltk.corpus.mac_morpho.sents()
text = "<p>"
for s in sents[:10]:
  text += " ".join(s)
  text += ".</p> |<p>"

print(len(text))
print(text[:1000])
````

````python
sents
````

````python
text = text.replace("<p>", "").replace("</p>", "").replace("|", "")
````

````python
frases = [f.strip() for f in text.split(".") if f.strip()]
frases
````

````python
frases = [f + "." for f in frases]
````

````python
frases
````

````python
from nltk.tokenize import RegexpTokenizer
tokenizer = RegexpTokenizer(r'\w+')

lista_de_tokens = [tokenizer.tokenize(frase) for frase in frases]
````

````python
lista_de_tokens
````

````python
from nltk.corpus import stopwords

# Baixar as stopwords (se ainda não tiver baixado)
nltk.download('stopwords')

# Carregar a lista de stopwords em português
stopwords_pt = set(stopwords.words('portuguese'))

# Remover as stopwords de cada frase (lista de listas de tokens)
tokens_filtrados = []
for frase in lista_de_tokens:  # 'tokens' é uma lista de listas de tokens, ou seja, frases já tokenizadas
    filtrados = [token for token in frase if token.lower() not in stopwords_pt]
    tokens_filtrados.append(filtrados)

# Mostrar o resultado
for i, frase in enumerate(tokens_filtrados, start=1):
    print(f"Frase {i} sem stopwords:", frase)
````
