## 1) Quais são as top 3 campanhas que geraram mais eventos? Ordene pela quantidade de eventos (2,5 pontos)

```python
# ESCREVA SEU CÓDIGO AQUI
```

```python
%%time

df.groupby('campaignId')\
['timestamp']\
.count()\
.rename("events_count")\
.nlargest(3)\
.compute()
```

## 2) Qual campanha teve mais clicks? (2,5 pontos)

```python
# ESCREVA SEU CÓDIGO AQUI
```

```python
%%time

df[df['action'] == 'click']\
  .groupby('campaignId')\
  ['timestamp']\
  .count()\
  .rename("clicks_count")\
  .nlargest(1)\
  .compute()
```

## 3) Qual mês teve o maior total de eventos acumulado? (2,5 pontos)

```python
df_month = df.copy()
df_month['month'] = df_month['timestamp'].dt.month
df_month.head()
```

```python
%%time
df_month.groupby('month')\
['timestamp']\
.count()\
.rename("events_count")\
.nlargest(1)\
.compute()
```

## 4) Nas situações onde existe um evento de view seguido de um evento de click criados pelo mesmo usuário no mesmo anúncio e campanha, quais são os 5 pares de anúncio e campanha com menores médias de tempo entre os dois eventos (2,5 pontos)

```python
# ESCREVA SEU CÓDIGO AQUI
```

```python
views = df[df["action"] == "view"].rename(columns={"timestamp": "view_time"})
views.persist()
```

```python
clicks = df[df["action"] == "click"].rename(columns={"timestamp": "click_time"})
clicks.persist()
```

```python
views = views.sort_values("view_time")
clicks = clicks.sort_values("click_time")
```

```python
merged = dd.merge_asof(
    views,
    clicks,
    left_on="view_time",
    right_on="click_time",
    by=["user_id", "adId", "campaignId"],
    direction="forward"  # garante que o click venha depois do view
)
merged.head()
```

```python
merged.isnull().sum().compute()
```

```python
merged = merged.dropna(subset=["click_time"])
```

```python
merged["diff_seconds"] = (
    merged["click_time"] - merged["view_time"]
).dt.total_seconds()
```

```python
medias = merged.groupby(['user_id',"adId", "campaignId"])["diff_seconds"].mean()
medias.nsmallest(5).compute()
```
