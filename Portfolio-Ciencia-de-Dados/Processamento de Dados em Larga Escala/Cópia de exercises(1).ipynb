```python
# IMPORTS, CRIAÇÃO DO CLUSTER E CARREGAMENTO DOS DADOS AQUI
import dask.dataframe as dd

from dask.distributed import Client, LocalCluster
```

```python
cluster = LocalCluster()
client = Client(cluster)
client
```

```python
df = dd.read_csv(
    'drive/MyDrive/data/ad_action.csv',
    header=None,
    parse_dates=[0])
df.columns = ['timestamp', 'user_id', 'action', 'adId', 'campaignId']
df
```

```python
df.head()
```

## Resposta Atividade 1

Qual é o anúncio mais popular, ou seja, com mais eventos independente do action?

**Resposta:**  
adId_06

```python
# ESCREVA SEU CÓDIGO AQUI
```

```python
df_aid = df.groupby("adId")\
 ["action"]\
 .count()\
 .nlargest(1)\
 .rename("count")\
 .reset_index()

df_aid.compute()
```

## Resposta Atividade 2

Quantos clicks gerou a campanha mais popular?

**Resposta:**  
63983 clicks

```python
# ESCREVA SEU CÓDIGO AQUI
```

```python
df_campaign = df[df["action"] == "click"]\
 .groupby("campaignId")\
 ["action"]\
 .count()\
 .nlargest(1)\
 .rename("count")\
 .reset_index()

df_campaign.compute()
```

## Resposta Atividade 3

Algum usuário só visualizou? Quantas actions foram enviadas por usuário que só visualizou?

**Resposta:**  
Sim, o usuário d99871cb-98b7-4ac5-97a5-b9a26c0f897b enviou apenas 1 action

```python
# ESCREVA SEU CÓDIGO AQUI
```

```python
result_df = df.groupby('user_id')\
  ['action']\
  .value_counts()\
  .rename('events_count')\
  .reset_index()\
  .categorize(columns=['action'])\
  .pivot_table(index='user_id', columns='action', values='events_count', aggfunc='sum')\
  .persist()
result_df.compute()
```

```python
result_df[result_df['click']==0].compute()
```
