# Sobre o Dataset

Este conjunto de dados, Student Mental Health Survey, é um conjunto de dados encontrado no Kaggle, focado no bem-estar estudantil, saúde mental e sucesso acadêmico, sendo uma ferramenta valiosa para pesquisadores e formuladores de políticas que buscam entender os desafios enfrentados por estudantes universitários. Ele contém informações detalhadas sobre diversos aspectos da vida dos estudantes, distribuídos em várias categorias:

Detalhes demográficos: Incluem dados como gênero, idade e a universidade de cada aluno.

Informações acadêmicas: Abrangem o nível de graduação, curso, ano acadêmico e o atual Índice de Aproveitamento Acadêmico.

Status residencial e experiências no campus: Registra o status residencial dos alunos e suas experiências com discriminação, assédio ou bullying no campus.

Fatores de estilo de vida: Incluem a frequência com que os alunos se envolvem em atividades esportivas e a média de horas de sono por noite.

Satisfação acadêmica: Coleta dados sobre o grau de satisfação dos alunos com seu curso e suas percepções sobre a carga de trabalho acadêmico.

Pressões e preocupações: Explora como os estudantes lidam com pressões acadêmicas, preocupações financeiras e a qualidade das relações sociais no ambiente universitário.

Saúde mental: Registra a frequência com que os alunos relatam sentir depressão, ansiedade, isolamento e insegurança em relação ao futuro.

Estratégias de enfrentamento: Documenta as atividades que os estudantes praticam para aliviar o estresse, proporcionando uma visão ampla de seus comportamentos e estratégias de autocuidado.

# Filtrando dados

**Análise da saúde mental por universidade**

Selecionando as 4 faculdades com mais aluno que apresentam dados mais significativos

```python
top_universidades = dataframe_saude_mental["university"].value_counts().head(4)
top_universidades
```

```python
index_universidades = top_universidades.index.tolist()
index_universidades
```

```python
data_universidades = dataframe_saude_mental[dataframe_saude_mental["university"].isin(index_universidades)]
data_universidades
```

```python
mean_values = data_universidades.groupby('university')[['depression', 'anxiety', 'isolation']].mean().sort_values(by='depression', ascending=False)
mean_values
```

```python
plt.figure(figsize=(10, 6))
mean_values.plot(kind='bar')
plt.title('Média de Saúde Mental por Universidade')
plt.xlabel('Universidade')
plt.ylabel('Média de Saúde Mental')
plt.xticks(rotation=0)
plt.legend(title="Métricas de saúde mental")
plt.show()
```

Entre as 4 universidades com mais alunos podemos observar:

-Universidade UET:

A segunda universidade com mais alunos,apresenta os níveis mais altos de depressão (4.30), ansiedade (4.30) e isolamento (4.10). Isso pode indicar que os alunos dessa universidade enfrentam maiores desafios emocionais e sociais.

-Universidade COMSATS:

Os níveis de depressão (3.56) e ansiedade (3.89) são altos, mas não tão elevados quanto os da UET. O isolamento (3.22) também é significativo, mas inferior ao da UET.

-Universidade PU:

A universidade com mais alunoa, apresenta níveis mais baixos em todas as categorias, com a depressão em 3.07, ansiedade em 2.96 e isolamento em 3.16. Isso sugere que os alunos dessa instituição podem estar em uma situação relativamente melhor em comparação com UET e COMSATS.

-Universidade FAST:

Tem os níveis mais baixos em todas as categorias: depressão (2.33), ansiedade (2.83) e isolamento (2.50). Os alunos dessa universidade parecem enfrentar menos problemas emocionais e sociais.

Análise da saúde mental de acordo com engajamento no esporte

```python
mean_values_sports = dataframe_saude_mental.groupby("sports_engagement")[['depression',"anxiety","isolation"]].mean().sort_values(by= "depression", ascending= False)
mean_values_sports
```

```python
plt.figure(figsize=(10,6))
mean_values_sports.plot(kind="bar")
plt.title("Média de Saúde Mental por Engajamento no Esporte")
plt.xlabel("Engajamento no Esporte")
plt.ylabel("Média de Saúde Mental")
plt.xticks(rotation=0)
plt.legend(title="Métricas de saúde mental")
plt.show()
```

1. Maior depressão entre aqueles que não praticam esportes:

Os estudantes que não praticam esportes apresentam a maior média de depressão (3.57), indicando que a ausência de atividade física pode estar associada a níveis mais altos de depressão.
Isso é consistente com estudos que mostram que a falta de exercício físico pode estar correlacionada com pior saúde mental, incluindo maiores índices de depressão e ansiedade.

2. Engajamento esportivo moderado (4-6 vezes por semana) associado a menor depressão:

Estudantes que se engajam em atividades esportivas entre 4-6 vezes por semana têm a menor média de depressão (2.09). Isso sugere que um nível regular e equilibrado de prática esportiva tem um impacto positivo significativo na saúde mental.

A prática regular de esportes está associada à liberação de endorfinas, melhoria da autoestima, redução do estresse e melhor regulação do humor, o que pode explicar a relação entre maior engajamento esportivo e menores níveis de depressão.

3. Engajamento esportivo leve (1-3 vezes por semana) e depressão moderada:

Estudantes que praticam esportes de 1 a 3 vezes por semana têm uma média de depressão intermediária (3.04), o que sugere que mesmo uma atividade física moderada pode ajudar a reduzir os níveis de depressão, embora o efeito não seja tão forte quanto a prática mais frequente.

4. Depressão relativamente alta entre aqueles que praticam esportes 7+ vezes por semana:

Curiosamente, os estudantes que praticam esportes 7 ou mais vezes por semana apresentam uma média de depressão de 3.40, que é mais alta do que os que praticam de 4 a 6 vezes e até mais alta que os que praticam de 1 a 3 vezes.
Esse dado pode sugerir que a prática excessiva de esportes, ou possivelmente a dependência de exercício ou o uso de esportes como um mecanismo de fuga, pode estar relacionado a níveis mais elevados de estresse ou outros problemas subjacentes de saúde mental. Também pode ser que esses indivíduos estejam sobrecarregando seus corpos, o que pode resultar em exaustão física e mental.

5. Benefícios do equilíbrio na prática esportiva:

O dado mais revelador é que o equilíbrio parece ser a chave: praticar esportes de 4 a 6 vezes por semana parece oferecer os melhores benefícios para a saúde mental, resultando nos menores níveis de depressão.
Tanto a falta de exercício quanto o excesso de atividade física parecem estar associados a níveis mais altos de depressão, o que sugere que a moderação é importante.

# Conclusão

#Impacto do Gênero na Saúde Mental:

As mulheres apresentam níveis mais altos de depressão, ansiedade e isolamento em comparação aos homens. Isso sugere que as mulheres podem estar enfrentando desafios emocionais maiores nas universidades, o que pode exigir intervenções e suporte mais direcionados.

#Diferenças por Universidade:

As faculdades têm níveis variados de saúde mental, com alguns cursos (como Software Engineering) apresentando níveis significativamente mais altos de depressão e ansiedade. Isso pode refletir as demandas acadêmicas e a cultura de cada programa, destacando a necessidade de suporte específico para áreas mais exigentes.

#Status Residencial:

A diferença entre os dois grupos é pequena, o que sugere que o status residencial não tem um impacto muito significativo nos níveis de depressão, ansiedade e isolamento. Ambos os grupos relatam níveis moderados de depressão,ansiedade e isolamento.

#Influência do Curso de Graduação:

Há uma variação significativa nos níveis de saúde mental de acordo com o curso de graduação. Os alunos de Software Engineering apresentam os níveis mais altos de dificuldades emocionais, enquanto os de Data Science parecem estar lidando melhor com essas questões. Isso sugere que o tipo de formação acadêmica pode estar relacionado ao estresse percebido e à saúde mental.

#Benefícios do Engajamento em Esportes:

A prática regular de esportes, especialmente de 4 a 6 vezes por semana, está associada aos menores níveis de depressão. A ausência de prática esportiva está relacionada aos maiores níveis de depressão, o que destaca a importância da atividade física na promoção do bem-estar mental. No entanto, o excesso de esportes (7 ou mais vezes por semana) parece estar relacionado a níveis mais altos de depressão, sugerindo que a moderação é importante para colher os benefícios da atividade física.

Essas conclusões gerais ressaltam a complexidade da saúde mental entre os alunos universitários e a necessidade de uma abordagem integrada que aborde as diversas dimensões que impactam o bem-estar emocional.
