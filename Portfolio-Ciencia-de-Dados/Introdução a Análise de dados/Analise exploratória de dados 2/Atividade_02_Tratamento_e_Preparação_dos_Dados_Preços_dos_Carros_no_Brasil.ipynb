#Gráficos

```python
plt.figure(figsize=(10, 6))
carros_por_marca.plot(kind='bar', color='skyblue')
plt.title('Quantidade de Carros por Marca')
plt.xlabel('Marca')
plt.ylabel('Quantidade de Carros')
plt.xticks(rotation=0)
plt.show()
```

```python
plt.figure(figsize=(10, 6))
tipos_engrenagem.plot(kind='bar', color='lightcoral')
plt.title('Quantidade de Carros por tipos de engrenagem')
plt.xlabel('Tipo Engrenagem')
plt.ylabel('Quantidade de carros')
plt.xticks(rotation=0)
plt.show()
plt
```

```python
dados_2022 = dados[dados['year_of_reference'] == 2022]
dados_2022
```

```python
meses_2022 = dados_2022.groupby('month_of_reference')['avg_price_brl'].mean()
meses_2022
```

```python
month_order = {
    'January': 1,
    'February': 2,
    'March': 3,
    'April': 4,
    'May': 5,
    'June': 6,
    'July': 7,
    'August': 8,
    'September': 9,
    'October': 10,
    'November': 11,
    'December': 12}
```

```python
meses_2022 = meses_2022.sort_index(key=lambda x: x.map(month_order))
meses_2022
```

```python
plt.figure(figsize=(10, 6))
meses_2022.plot(kind='line', color='lightgreen',linewidth=3, marker= 'o')
plt.title('Preço Médio dos Carros por Mês em 2022')
plt.xlabel('Mês')
plt.ylabel('Preço Médio')
plt.xticks(rotation=45)
plt.grid()
plt.tight_layout()
plt.show()
```

```python
dados1 = dados.loc[:,['brand',"gear","avg_price_brl"]]
dados1
```

```python
dados_grafico1= dados1.groupby(['brand','gear'])['avg_price_brl'].mean()
dados_grafico1
```

```python
plt.figure(figsize=(10, 6))
dados_grafico1.plot(kind='bar', color='lightcoral')
plt.title('Preço Médio dos Carros por Marca e Tipo de Engrenagem')
plt.xlabel('Marca e Tipo de Engrenagem')
plt.ylabel('Preço Médio')
plt.xticks(rotation=45)
plt.grid()
plt.tight_layout()
plt.show()
```

Os resultados mostram que veículos automáticos são mais caros que manuais para todas as marcas analisadas. A Volkswagen possui o maior preço médio para automáticos (R\$ 99.735), enquanto a Fiat apresenta a maior diferença entre os preços de automáticos (R\$ 97.397) e manuais (R\$ 39.694). Renault é a marca com menor variação de preço entre esses tipos de engrenagem.

```python
dados2 = dados.loc[:,['brand',"fuel","avg_price_brl"]]
dados2
```

```python
dados_grafico2= dados2.groupby(['brand','fuel'])['avg_price_brl'].mean()
dados_grafico2
```

```python
plt.figure(figsize=(10, 6))
dados_grafico2.plot(kind='bar', color='lightcoral')
plt.title('Preço Médio dos Carros por Marca e Tipo de Combustível')
plt.xlabel('Marca e Tipo de Combustível')
plt.ylabel('Preço Médio')
plt.xticks(rotation=45)
plt.grid()
plt.tight_layout()
plt.show()
```

Os resultados mostram que veículos a diesel são os mais caros para todas as marcas, especialmente da Volkswagen (R\$ 139.216) e Renault (R\$ 104.530).Carros a álcool possuem os preços mais baixos, destacando-se Fiat (R\$ 11.509), Ford (R\$ 10.149) e Chevrolet (R\$ 13.698). Já os veículos a gasolina apresentam preços variando entre R\$ 37.197 (Fiat) e R\$ 59.043 (Nissan).
