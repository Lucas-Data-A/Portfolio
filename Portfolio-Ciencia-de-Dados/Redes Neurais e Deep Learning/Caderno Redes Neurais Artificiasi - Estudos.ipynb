### J (função de custo/perda): coeficiente utilizado na equação para atualizar os pesos da rede, o gradiente descendente.

```python
#Importações
%matplotlib inline
import matplotlib.pyplot as plt   # para plotting

from sklearn.datasets import load_iris  # para carregar dataset

import numpy as np  # processamento matricial

np.random.seed(1234)
```

```python
iris = iris = load_iris()
```

```python
iris
```

```python
data = iris.data[iris.target==1,::2]  # comprimento das sépalas e pétalas da versicolor(target = 1), indices 0 e 2 (sepal e petal length)
```

```python
#Separação dos dados
x_train = data[:,0:1]
y_train = data[:,1:2]
#Usando 0:1 e 1:2 (ao invés de só 0 e 1) para manter o formato de matriz 2D, necessário para muitos algoritmos de machine learning.
```

```python
n_samples = x_train.shape[0]
print('x_train.shape:',x_train.shape, x_train.dtype)
print('y_train.shape:',y_train.shape, y_train.dtype)

print('x_train[:5]:\n', x_train[:5])
print('y_train[:5]:\n', y_train[:5])

#Exibe o formato (shape) e tipo de dados (dtype) de x_train e y_train.
#Mostra as 5 primeiras amostras de x_train e y_train.
```

```python
#Normalização dos dados
x_train -= x_train.min()
x_train /= x_train.max()
y_train -= y_train.min()
y_train /= y_train.max()

# normalização max-min scale
# Resultado: agora o menor valor de x_train é 0.
# Resultado: agora o maior valor de x_train vira 1.
```

```python
#Plot de gráfico:
plt.scatter(x_train,y_train)
plt.xlabel('Comprimento sepalas')
plt.ylabel('Comprimento petalas')
```

```python
x_train
```

```python
#Modelando a rede:
#Preenchendo x com coluna de 1's para bias
x_train_bias = np.hstack([np.ones(shape=(n_samples,1)), x_train])
```

```python
x_train_bias
```

```python
### Classe com o modelo da rede
class Net():
    def __init__(self, n_in, n_out):
        self.w = np.random.uniform(-0.1,0.1,(n_out,n_in)) #inicialização dos parâmetros

    def forward(self, x_bias):
        return x_bias.dot(self.w.T) #^y

#__init__ é o construtor da classe — chamado automaticamente quando você cria uma instância de Net.
# n_in = número de entradas (features de entrada)
# n_out = número de saídas (quantos valores a rede devolve)
# self.w É uma matriz de pesos (n_out linhas, n_in colunas).Inicializada com valores gerados aleatoriamente entre -0.1 e 0.1.Essa matriz será usada para calcular as saídas.
# Cria os pesos iniciais da rede, aleatórios pequenos.
# Calcula as previsões multiplicando entrada × pesos.
```

visualização prática.PNG

```python
model = Net(2,1) # duas entradas (1 + x0) e uma saída y_pred
```

```python
### Testando o predict da rede (mesmo sem treiná-la)
xin = np.random.randn(4,2) # 4 amostras, 2 entradas (1 + x0)
ypred = model.forward(xin)
ypred


```

lembrando.PNG

atualização de pesos.PNG

```python
# Laço de treinamento
num_epochs = 100 #número de épocas
learning_rate = 0.5
for epoch in range(num_epochs):

    # forward - predict (foward cálcula os valores de y pred)
    y_pred = model.forward(x_train_bias)

    #loss cálculo da função de perda (MSE)
    loss = np.square(y_pred - y_train).mean()

    # cálculo do gradiente pelas derivadas parciais (o triângulo J da função de perda)
    w_grad = (2.0/n_samples) * (x_train_bias.T).dot(x_train_bias.dot(model.w.T) - y_train)
#                             (x_train_bias.T).dot(...) = produto escalar da matriz transposta de entrada com o vetor de erros  x_train_bias.dot(model.w.T) - y_train =  (y^ - y)
    # gradiente descendente
    model.w = model.w - learning_rate * w_grad.T

    # verbose
    if (epoch+1) % 1 == 0:
        print('Epoch[{}/{}], loss: {:.6f}'
              .format(epoch+1, num_epochs, loss))
```

# Regressão Linear com PyTorch

```python
%matplotlib inline
import torch

import matplotlib.pyplot as plt
from sklearn.datasets import load_iris

torch.manual_seed(1234)

def print_assert(r):
    if r:
        return 'OK!'
    else:
        return 'Incorreto!'
```

```python
import pandas as pd
iris = load_iris()
data = iris.data[iris.target==1,::2]  # comprimento das sépalas e pétalas, indices 0 e 2

x_in = data[:,0:1]
y_in = data[:,1:2]

x_train = torch.FloatTensor(x_in) #A torch.Tensor is a multi-dimensional matrix containing elements of a single data type.
y_train = torch.FloatTensor(y_in) # uma estrutura de dados geral (como um array ou matriz) usada para representar dados numéricos multidimensionais (escalares, vetores, matrizes ou tensores de ordens superiores).

n_samples = x_train.size(0) #só o primeiro valor
print('x_train.shape:',x_train.shape, type(x_train))
print('y_train.shape:',y_train.shape, type(y_train))

print('x_train[:5]:\n', x_train[:5])
print('y_train[:5]:\n', y_train[:5])

iris_pd = pd.DataFrame(x_in, columns=['x_in'])
iris_pd['y_in'] = y_in
iris_pd.head()
```

```python
#Normalização
x_train -= x_train.min()
x_train /= x_train.max()
y_train -= y_train.min()
y_train /= y_train.max()

plt.scatter(x_train,y_train)
plt.xlabel('Comprimento sepalas')
plt.ylabel('Comprimento petalas')
```

```python
#Criação do `x_train_bias`
# faça o `x_train_bias` ser uma matriz com 1s na primeira coluna e x_train na segunda coluna.
ones = (torch.ones(n_samples,1))
ones.shape
```

```python
ones
```

```python
x_train.shape
```

```python
x_train
```

```python
x_train_bias = torch.cat((ones,x_train),1)
x_train_bias
```

```python
# Verificação do exercício - Execute para testar sua implementação

print('x_train_bias é um tensor: {}'.format(print_assert(
    isinstance(x_train_bias, torch.FloatTensor)
)))

print('tamanho do x_train_bias: {}'.format(print_assert(
    x_train_bias.size() == torch.Size([50, 2])
)))

print('primeira coluna é só de uns: {}'.format(print_assert(
    (x_train_bias[:, 0] - torch.ones(n_samples)).sum() == 0
)))

print('segunda coluna é igual a x_train: {}'.format(print_assert(
    (x_train_bias[:, 1] - x_train.reshape(50)).sum() == 0
)))
```

```python
#Modelo Rede:
#Definição da classe
class Net():
    def __init__(self, n_in, n_out):
        self.w = torch.Tensor(n_out, n_in) # Cria self.w, um tensor de pesos com shape (n_out, n_in)
        self.w.uniform_(-0.1,0.1) #inicialização dos parâmetros(pesos)

    def forward(self, x_bias):
        return x_bias.matmul(torch.t(self.w)) # O .matmul() no PyTorch faz a multiplicação matricial (ou produto de matrizes) entre dois tensores, de acordo com as regras da álgebra linear.
```

```python
#Iniciando a rede:
model = Net(2,1)
```

```python
# Testando o predict da rede:
model.forward(torch.ones((4,2)))
#model.forward(torch.randn(4,2))
```

```python
# Treinamento
num_epochs = 100
learning_rate = 0.5
for epoch in range(num_epochs):

    # forward - predict
    y_pred = model.forward(x_train_bias)

    #loss cálculo da função de perda
    #loss = np.square(y_pred - y_train).mean()
    loss = torch.mean(torch.square(y_pred - y_train))
    #loss = torch.mean((y_pred - y_train) ** 2)

    # cálculo do gradiente pelas derivadas parciais
    #w_grad = (2.0/n_samples) * (x_train_bias.T).dot(x_train_bias.dot(model.w.T) - y_train)
    w_grad = (2.0/n_samples) * torch.matmul((torch.t(x_train_bias)), (torch.matmul(x_train_bias, torch.t(model.w)) - y_train))

    # gradiente descendente
    #model.w = model.w - learning_rate * w_grad.T
    model.w = model.w - learning_rate * torch.t(w_grad)

    # verbose
    if (epoch+1) % 20 == 0:
        print('Epoch[{}/{}], loss: {:.6f}'
              .format(epoch+1, num_epochs, loss))
```

```python
# Avaliação:
#Predict:
y_pred = model.forward(x_train_bias)
plt.plot(x_train.numpy(), y_train.numpy(), 'ro', label='Original data')
plt.plot(x_train.numpy(), y_pred.numpy(), 'kx-', label='Fitting Line')
plt.show()
```

# PyTorch: Variable, Gradientes e Grafo Computacional

Este notebook introduz
- o conceito de `Variables` do PyTorch,
- uma interpretação numérica intuitiva do gradiente
- grafo computacional, utilizado para o cálculo automático do gradiente de uma função.

Um dos principais fundamentos para que o PyTorch seja adequado para deep learning é a sua habilidade de
calcular o gradiente automaticamente a partir da expressões definidas. Essa facilidade é implementada
pelo tipo Variable do PyTorch, que adiciona ao tensor a facilidade de cálculo automático do gradiente pela construção dinâmica do grafo computacional.

**Variable** (obsoleto, mas historicamente importante)
Antigamente, o PyTorch usava uma classe chamada Variable para envolver tensores que precisavam de gradientes. A partir do PyTorch 0.4.0, Variable foi fundido com Tensor, ou seja, todo Tensor agora pode ter gradientes.

Hoje, você simplesmente usa torch.Tensor com requires_grad=True.

Um tensor com gradiente é um tensor que o PyTorch está monitorando para calcular derivadas (gradientes) automaticamente em relação a alguma função de perda ou saída escalar.

**Gradientes** representam a derivada da função de perda em relação a um tensor (geralmente os pesos de uma rede neural). Se você quiser que um tensor tenha gradiente, você precisa definir requires_grad=True.

Durante o treinamento:

Você realiza operações com tensores.

Chama .backward() no valor final (ex: função de perda).

O PyTorch preenche .grad com os gradientes.

O PyTorch usa **um grafo dinâmico** para calcular os gradientes. Isso significa que o grafo é construído em tempo de execução, com base nas operações feitas com os tensores que têm requires_grad=True.

Cada operação cria nós no grafo, que apontam para os tensores anteriores, formando um grafo direcionado. Quando você chama .backward(), o PyTorch faz a propagação reversa (backpropagation) nesse grafo.

✔ Vantagens do grafo dinâmico:

Fácil de depurar (você pode usar print, if, for, etc.).

Flexível (ideal para modelos que mudam de forma, como RNNs com diferentes comprimentos de sequência).



J.PNG

Variable possui 3 campos: o dado em si (data), o gradiente (grad) e um apontador (creator) para construir o grafo da backpropagation. Uma expressão utilizada para o cálculo do gradiente exige que todas suas expressões sejam calculadas com Variables, caso contrário não é possível construir o grafo computacional.

![alt text](https://raw.githubusercontent.com/vcasadei/images/master/variables.png)

No PyTorch antigo (Variable), cada objeto tinha:

**data**: o tensor bruto (os números que você manipula, por exemplo [1.0, 2.0]).

**grad**: o gradiente calculado durante a backpropagation (por exemplo, a derivada da perda em relação a esse tensor).

**creator (hoje grad_fn)**: um ponteiro para a função que criou esse tensor — isso é o que liga os nós do grafo computacional e permite fazer a propagação reversa.

```python
#  Nota moderna: Em versões atuais do PyTorch, não é mais necessário usar Variable. Basta:
w = torch.ones(1, requires_grad=True);w
```

Seja a expressão: $$ J = ((x  w) - y)^2 $$

Queremos calcular a derivada de $J$ em relação a $w$.

```python
## Cálculo automático do gradiente da função perda J
### Montagem do grafo computacional
# predict (forward)
y_pred = x * w # previsões com os dados x para os pesos w

# cálculo da perda J: loss
e = y_pred - y # o erro individual de uma interação
e2 = e.pow(2) # o erro ao quadrado(MSE)
J = e2.sum() # O somatório dos erros, o que origina J
J #Função de perda
```

```python
## Auto grad - processa o grafo computacional backwards
```

O `backward()` varre o grafo computacional a partir da variável a ele associada e calcula o gradiente para todas as `Variables` que possuem o atributo `requires_grad` como verdadeiro.
O `backward()` destroi o grafo após sua execução. Isso é intrínsico ao PyTorch pelo fato dele ser uma rede dinâmica.

```python
J.backward()
print(w.grad)
```

No PyTorch, os gradientes são acumulados a cada chamada de backward(). Isso é útil para alguns algoritmos (como otimização por minibatches), mas na prática você quase sempre precisa zerá-los antes de cada novo passo de otimização.

## Interpretação do Gradiente

O gradiente de uma variável final (J) com respeito à outra variável de entrada (w) pode ser interpretado como o quanto a variável final J vai aumentar se houver um pequeno aumento na variável de entrada (w). Por exemplo suponha que o gradiente seja 28. Isto significa se aumentarmos a variável w de 0.001, então J vai aumentar de 0.028.

```python
eps = 0.001
y_pred = x * (w+eps)
J_new = (y_pred - y).pow(2).sum()
J_new
```

## Back propagation
Uma forma equivalente explícita de calcular o gradiente é fazendo o processamento do backpropagation no grafo computacional, de forma explícita.
Apenas como ilustração.

Função de perda:
$$ J(\hat{y_i},y_i) = \frac{1}{M} \sum_{i=0}^{M-1} (\hat{y_i} - y_i)^2 $$

Gradiente:
$$  \mathbf{\nabla{J_w}} = \frac{2}{M}\mathbf{x^T}(\mathbf{x w^T} - \mathbf{y}) $$

Atualização dos parâmetros pelo gradiente descendente:
$$ \mathbf{w} = \mathbf{w} − \eta (\mathbf{\nabla J_w})^T $$


```python
import numpy as np

dJ = 1.
de2 = dJ * np.ones((4,))
# Calcula o gradiente de J = e2.sum() com relação a e2
# Como J = e2[0] + e2[1] + e2[2] + e2[3], a derivada de J com respeito a cada elemento de e2 é 1
de = de2 * 2 * e.data.numpy() # e = [0,-1,-2,-3]
dy_pred = de # de = [0,-2,-4,-6]
dw = (dy_pred * x.data.numpy()).sum() #[0,-2,-4,-6]*[0,1,2,3] = [0,-2,-8,-18].sum() = -28
print(dJ)
print(de2)
print(de)
print(dw)
```

```python
#1.
# predict (forward)
w = Variable(torch.ones(1),requires_grad=True); w
y_pred = x * w

# cálculo da perda J: loss
e = y_pred - y
e2 = e.pow(2)
J = e2.sum()
J
print(f"J antigo: {J}")

J.backward()

lr = 0.1;
w = w - lr * w.grad.data


print(w)

y_pred = x * (w)
J_new = (y_pred - y).pow(2).sum()
J_new
print(f"J novo: {J_new}")
```

```python
#Segundo o gpt
w = torch.tensor([1.0], requires_grad=True)

for i in range(10):
    y_pred = x * w
    J = ((y_pred - y)**2).sum()

    J.backward()

    print(f"Iteração {i} | J: {J.item()} | w: {w.item()} | grad: {w.grad.item()}")

    # Atualização manual
    with torch.no_grad():  # evita que a atualização seja rastreada no grafo
        w -= lr * w.grad

    w.grad.zero_()  # limpa gradiente acumulado
```
