```python
import os
import shutil
import random
from sklearn.model_selection import train_test_split
```

Dataset base do kaggle:
https://www.kaggle.com/datasets/andrewmvd/animal-faces

Este conjunto de dados, também conhecido como Animal Faces-HQ (AFHQ), consiste em 16.130 imagens de alta qualidade com resolução de 512×512.
Há três domínios de classes, cada um contendo cerca de 5.000 imagens.
As classes são:

* Gato;

* Cachorro;

* Animais Selvagens.

Para melhorar o desempenho e acelerar o processo de treinamento,
estamos criando um subconjunto menor do dataset AFHQ, reduzindo a quantidade de imagens por classe tanto no treino como na validação.

```python
# Define o caminho para o diretório onde está o dataset AFHQ dentro do Google Drive
from google.colab import drive
drive.mount('/content/drive')
orig_dir = '/content/drive/My Drive/Colab Notebooks/afhq'
```

```python
# Define o caminho onde será salvo o dataset reduzido
reduced_dir = '/content/drive/My Drive/Colab Notebooks/Animais'
# Cria o diretório se ele ainda não existir. O parâmetro 'exist_ok=True' evita erro caso o diretório já exista
os.makedirs(reduced_dir, exist_ok=True)
```

```python
# Parâmetros que queremos no Dataset subconjunto criado
train_count = 50 # Número de imagens para o conjunto de treino por classe
val_count = 50 # Número de imagens para o conjunto de validação por classe
total_required = train_count + val_count # Total de imagens necessárias por classe
splits = ['train', 'val']  # Subpastas do dataset original que contêm as imagens

# Para cada classe
classes = ['cat', 'dog', 'wild']
for class_name in classes:
    # Coletar imagens das pastas train e val da classe
    all_images = []
     # Coleta imagens tanto da pasta 'train' quanto da 'val'
    for split in splits:
        class_dir = os.path.join(orig_dir, split, class_name)  # Caminho para a pasta da classe
        if not os.path.isdir(class_dir): # Se a pasta não existir, pula
            continue
        imgs = [f for f in os.listdir(class_dir) if f.lower().endswith('.jpg')] # Filtra apenas arquivos com extensão .jpg (ignora outros tipos)
        all_images += [os.path.join(class_dir, f) for f in imgs] # Adiciona o caminho completo das imagens à lista

    # Verifica se há imagens suficientes para essa classe
    if len(all_images) < total_required:
        print(f"❌ Classe '{class_name}' tem apenas {len(all_images)} imagens. Pulando...")
        continue

    # Selecionar aleatoriamente 75 imagens
    random.seed(42)
    selected = random.sample(all_images, total_required) # Seleciona as imagens aleatoriamente

    # Dividir as imagens selecionadas em treino e validação
    train_imgs, val_imgs = train_test_split(
        selected, test_size=val_count, train_size=train_count, random_state=42
    )

    # Copiar para a nova estrutura
    for split_name, img_list in zip(['train', 'val'], [train_imgs, val_imgs]):
        dest_dir = os.path.join(reduced_dir, split_name, class_name)  # Caminho do destino
        os.makedirs(dest_dir, exist_ok=True)  # Cria a pasta de destino, se não existir

        for img_path in img_list:
            fname = os.path.basename(img_path) # Pega apenas o nome do arquivo
            dst = os.path.join(dest_dir, fname)  # Caminho de destino final do arquivo
            shutil.copyfile(img_path, dst) # Copia o arquivo para o novo destino

# Mensagem final confirmando que o processo foi concluído
print("✅ Dataset reduzido e organizado salvo na pasta: Animais")
```
