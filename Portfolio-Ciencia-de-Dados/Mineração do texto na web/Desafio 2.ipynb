```python
from sklearn.datasets import fetch_20newsgroups

#Descomente a linha de baixo se quiser ver mais detalhes do dataset
print(fetch_20newsgroups.__doc__)
N = 15
raw_documents= fetch_20newsgroups(subset='train', return_X_y=True)[0][:N]
for idx, doc in enumerate(raw_documents[:N]):
  print(f"========== Doc {idx} ==========")
  print(doc)
```

```python
from sklearn.feature_extraction.text import CountVectorizer
import nltk
nltk.download('stopwords')
from nltk.corpus import stopwords
```

```python
vectorizer = CountVectorizer(lowercase=True,stop_words=stopwords.words('english'), max_df=0.8, min_df=1)
vectorizer.fit(raw_documents)
```

```python
sorted(vectorizer.vocabulary_.items(), reverse = False)
```

```python
terms = vectorizer.get_feature_names_out()
print(terms)
```

```python
X_counts = vectorizer.fit_transform(raw_documents)
print(X_counts[0,:]) #TF
```

```python
X_counts.shape
```

```python
from tkinter.constants import X
from sklearn.feature_extraction.text import TfidfTransformer

transformer = TfidfTransformer()
X_tfidf = transformer.fit_transform(X_counts)
idf_values = transformer.idf_
print(X_tfidf[0,:])
```

```python
import pandas as pd
```

```python
rows = []
for doc_idx in range(X_counts.shape[0]):
    for term_idx, term in enumerate(terms):
        tf = X_counts[doc_idx, term_idx]
        tfidf = X_tfidf[doc_idx, term_idx]
        if tf > 0:  # Só inclui termos que aparecem no documento
            rows.append({
                'Documento': doc_idx,
                'Termo': term,
                'TF': tf,
                'IDF': idf_values[term_idx],
                'TF-IDF': tfidf
            })

df = pd.DataFrame(rows)
```

```python
print(df)
```

```python
top5 = df.sort_values(by='TF-IDF', ascending=False).head(5)
print(top5)
```
