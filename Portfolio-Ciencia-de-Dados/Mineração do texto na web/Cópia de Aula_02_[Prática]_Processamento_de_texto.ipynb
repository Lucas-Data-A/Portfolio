## 04.Tokenização

````python
from nltk.corpus import stopwords

tokenizer = RegexpTokenizer(r'\w+')
stop_words = stopwords.words('portuguese')
````

Obs: Decidimos optar pela remoção dos números, pois ficavam muito soltos dentro do contexto!!

Além disso decidimos tratar as sentenças separadamente, aplicando o set separadamente para cada frase para que não se perdessem palavras essenciais para cada contexto.

````python
import re
tokens = []
pre_list = []
dict = {}
#stop_words.remove("são")
for n, sentence in enumerate(list_sentences):
  dict[n] = sentence
for key,value in dict.items():
    pre_list += value.lower().replace(",", "").split(" ")
    frase = [dict.fromkeys([s.lower() for s in tokenizer.tokenize(value)])]  # Removendo redundância nas palavras porém mantendo a ordem da frase
    for v in frase:
        v = [token for token in v if not token in stop_words]  # Removendo Stopwords
        tokens += v
        for n in tokens:
          if n.isdigit():
            tokens.remove(n)
print(pre_list)
print(tokens)
print(f"Total de palavras depois do Set & StopWords:{len(pre_list)}")
print(f"Total de palavras depois do Set & StopWords:{len(tokens)}")
````

# 05. Lemmatization x Stemming

````python
lemma_words = [wordnet_lemmatizer.lemmatize(word) for word in tokens]
stem_words = [porter_stemmer.stem(word) for word in tokens]


print(f"Quantidade de palavras usando Lemmatization:{len(lemma_words)}")
print(f"Quantidade de palavras usando Stemming:{len(stem_words)}")

print(lemma_words)
print(stem_words)

print("---Lemmatization---")
for n, value in enumerate(tokens):
    if value == lemma_words[n]:
        pass
    else:
        print(f"A palavra '{value}' foi alterada para '{lemma_words[n]}'")

print("---Stemming---")
for n, value in enumerate(tokens):
    if value == stem_words[n]:
        pass
    else:
        print(f"A palavra '{value}' foi alterada para '{stem_words[n]}'")
````

````python
df = pd.DataFrame(data=[np.zeros(len(list(set(tokens))))], columns=list(set(tokens)), index=["N","TF"])
quant_token = len(list(set(tokens)))
````

````python
for word in pre_list:
  if word in df.keys():
    df[word]["N"] += 1

for token in df.keys():
  df[token]["TF"] = df[token]["N"]/quant_token

df
````
