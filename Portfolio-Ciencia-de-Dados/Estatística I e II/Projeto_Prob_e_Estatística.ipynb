#**Probabilidade Simples**

Qual a probabilidade de selecionar um filme aleatório da lista e esse filme ser um Documentário?

```python
probDo = len(filmes[filmes['Gênero']=='Documentário'])/len(filmes)

print('A probabilidade de selecionar um filme aleatório da lista e esse filme ser um Documentário é de: ' + str(round(probDo*100,2))+'%')
```

```python
#referência externa: https://letterboxd.com/films/by/rating/

probDoLe = 176486/997089
print('A probabilidade de escolher um filme aleatório no site Letterboxd e ele ser documentário é: ' + str(round(probDoLe*100,2))+'%')
```

#**Probabilidade da União**

Qual a probabilidade de selecionar um filme aleatório da lista e esse filme ser um Documentário ou uma Animação?

```python
doA = len(filmes[(filmes['Gênero']=='Documentário') | (filmes['Gênero']=='Animação')])
probDoA = doA/len(filmes)
print('A probabilidade de selecionar um filme aleatório da lista e esse filme ser um Documentário ou uma Animação é de: ' + str(round(probDoA*100,2))+'%')
```

#**Probabilidade da Intersecção**

Qual a probabilidade de ser um Documentário e ser nacional?

```python
doN = len(filmes[(filmes['Gênero']=='Documentário') & (filmes['Nacionalidade da obra']=='Brasileira')])
probDoN = doN/len(filmes)
print('A probabilidade de ser um Documentário e ser nacional é de: ' + str(round(probDoN*100,2))+'%')
```

#**Probabilidade do Complemento**

Qual a probabilidade de ser um Documentário e ser internacional?

```python
probNDoN = probDo - probDoN
print('A probabilidade de ser um Documentário e ser internacional é de: ' + str(round(probNDoN*100,2))+'%')
```

#**Probabilidade da Diferença**

Qual a probabilidade de ser do conjunto internacional mas não ter de pais produtor Estados Unidos?

```python
iNEUA = len(filmes[(filmes['Nacionalidade da obra']!='Brasileira') & (filmes["País(es) produtor(es) da obra"]!='Estados Unidos') ])
probI = iNEUA/len(filmes)
print('A probabilidade de ser do conjunto internacional mas não ter de pais de produtor Estados Unidos é de: ' + str(round(probI*100,2))+'%')
```

#**Probabilidade Condicional**

Qual a probabilidade de ser nacional sabendo que é um documentário?

```python
probCond = probDoN/probDo
print('A probabilidade de ser nacional sabendo que é um documentário é de: ' + str(round(probCond*100,2))+'%')
```

#**Teorema de Bayes**

Qual a probabilidade de ser um documentário talque seja nacional?

```python
#Probabilidade do filme ser nacional
probN = len(filmes[filmes['Nacionalidade da obra']=='Brasileira'])/len(filmes)
```

```python
probDotqN = (probDoN*probCond)/probN
print('A probabilidade de ser um documentário talque seja nacional é de: ' + str(round(probDotqN*100,2))+'%')
```

##**Função Probabilidade**

```python
fp = filmes['Ano de exibição'].value_counts()/len(filmes)
fp = fp.sort_index()
print(fp)
```

##**Função Repartição**

```python
fp.cumsum()
```

##**Gráfico Simples**

```python
import matplotlib.pyplot as plt
plt.plot(fp, '*-')
plt.xticks([2015,2016,2017,2018,2019])

plt.title('Função de Probabilidade')
plt.ylabel('Probabilidade')
plt.xlabel('Ano')

plt.show()
```

```python
plt.plot(fp.cumsum(), '.-', color='red')
plt.xticks([2015,2016,2017,2018,2019])

plt.title('Função Repartição')
plt.ylabel('Probabilidade Acumulada')
plt.xlabel('Ano')

plt.show()
```
