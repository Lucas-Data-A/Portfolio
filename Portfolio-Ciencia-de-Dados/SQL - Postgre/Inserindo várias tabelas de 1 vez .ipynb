```python
import pandas as pd
```

```python
from sqlalchemy import create_engine
```

```python
#FORMA MANUAL
```

```python
Produto = pd.read_excel(r"E:\ETL\Origem\arquivos_excel\Produto.xlsx")
```

```python
Produto.head()
```

```python
Categoria  = pd.read_excel(r"E:\ETL\Origem\arquivos_excel\Categoria.xlsx")
```

```python
Categoria.head()
```

```python
engine = create_engine("postgresql+psycopg2://postgres:Naruto120@localhost:5432/testes")
```

```python
tabelas = {
    "produtos": Produto,
    "categorias": Categoria
}
```

```python
for nome_tabela, df in tabelas.items():
    df.to_sql(
        name=nome_tabela,
        con=engine,
        if_exists="replace",  # ou "append", se quiser acrescentar
        index=False
    )
    print(f"Tabela {nome_tabela} carregada com sucesso!")
```

```python
pasta = r"E:\ETL\Origem\arquivos_excel"
```

```python
import os
```

```python
for arquivo in os.listdir(pasta):
    if arquivo.endswith(".xlsx"):
        nome_tabela = arquivo.replace(".xlsx", "").lower()
        df = pd.read_excel(os.path.join(pasta, arquivo))
        df.to_sql(nome_tabela, engine, if_exists="replace", index=False)
        print(f"Tabela {nome_tabela} criada a partir de {arquivo}")
```
