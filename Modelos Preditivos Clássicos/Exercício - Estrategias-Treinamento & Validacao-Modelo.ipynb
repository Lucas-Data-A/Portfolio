```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

## Obtendo os dados utilizando make_classification

```python
from sklearn.datasets import make_classification

X, y = make_classification(
    n_samples=1000,
    n_features=20,
    n_informative=3,
    n_redundant=0,
    n_repeated=0,
    n_classes=2,
    random_state=0,
    shuffle=False,
)
```

```python
pd.DataFrame(X)
```

```python
pd.DataFrame(y)
```

# Divisão treino-teste

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y,
                                                    test_size=0.3,
                                                    random_state=111)
```

```python
pd.DataFrame(X_train)
```

# PCA (treinando modelo com 10 e 5 features)

```python
from sklearn.decomposition import PCA
```

```python
n_components = 10
pca10 = PCA(n_components=n_components)
```

```python
X_train_pca10 = pca10.fit_transform(X_train)
X_test_pca10 = pca10.transform(X_test)
```

```python
pd.DataFrame(X_train_pca10)
```

```python
n_components = 5
pca5 = PCA(n_components=n_components)
```

```python
X_train_pca5 = pca5.fit_transform(X_train)
X_test_pca5 = pca5.transform(X_test)
```

```python
pd.DataFrame(X_train_pca5)
```

# Analisando resultados dos experimentos

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import classification_report
```

```python
knn = KNeighborsClassifier(n_neighbors=5)
knn.fit(X_train, y_train)
y_pred = knn.predict(X_test)
```

```python
print(classification_report(y_test, y_pred))
```

```python
knn = KNeighborsClassifier(n_neighbors=5)
knn.fit(X_train_pca10, y_train)
y_pred = knn.predict(X_test_pca10)
```

```python
print(classification_report(y_test, y_pred))
```

```python
knn = KNeighborsClassifier(n_neighbors=5)
knn.fit(X_train_pca5, y_train)
y_pred = knn.predict(X_test_pca5)
```

```python
print(classification_report(y_test, y_pred))
```

# Usando o algoritmo Random Forest como Feature Selection

A utilização  das abordagens de PCA, Feature Selection e Validação Cruzada tiveram um impacto na otimização do desempenho dos modelos criados durante o exercício. A utilização dessas técnicas possibilitou um processo de validação e pré-processamento mais eficaz, resultando em um aumento da precisão dos modelos finais.

A utilização do PCA possibilitou uma diminuição da dimensionalidade dos dados, removendo redundâncias e correlações exageradas entre atributos, o que resultou em modelos mais leves e com superior capacidade de generalização. Essa compressão dos dados foi executada com a menor perda de variância explicativa, preservando as informações mais importantes para o treinamento do algoritmo.

A técnica de Feature Selection, por sua vez, ajudou a aumentar a eficiência e a interpretabilidade do modelo, selecionando apenas os atributos mais significativos para a tarefa de classificação. Isso diminuiu o ruído nos dados e favoreceu uma modelagem mais exata, ao impedir que variáveis irrelevantes afetassem os resultados de forma negativa.

Além disso, a aplicação da Cross Validation foi fundamental para assegurar a credibilidade das métricas de avaliação. A segmentação sistemática dos dados em vários subconjuntos possibilitou uma avaliação mais sólida e realista do desempenho dos modelos, evitando questões como overfitting e assegurando que os aumentos de precisão verificados fossem consistentes.

Assim, a utilização dessas três estratégias no processo de modelagem não apenas elevou a precisão dos modelos preditivos, mas também auxiliou na criação de soluções mais eficazes, compreensíveis e resilientes. Os resultados alcançados evidenciam a relevância dessas abordagens no processo de criação de modelos de machine learning de excelente qualidade.
